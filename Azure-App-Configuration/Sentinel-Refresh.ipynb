### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()
app_config_conn_string = os.getenv("AZURE_APPCONFIG_CONNECTION_STRING")
# configuration plans in the 
```

### Creating the App Configuration Client

```python
from azure.appconfiguration.provider import load, WatchKey

# creating the Azure App Configuration loader object
config = load(
    connection_string=app_config_conn_string,
    refresh_on = [WatchKey("Sentinel")],
    refresh_interval = 30
)
```

### Implement the Refresh Strategy

```python
config.refresh()

print(f"Azure OpenAI Endpoint: {config["AzureOpenAI:Endpoint"]}")
print(f"Azure OpenAI API Key: {config["AzureOpenAI:Key"]}")
print(f"Azure OpenAI Model Name: {config["AzureOpenAI:ModelName"]}")
```

```python
config.refresh()

print(f"Azure OpenAI Endpoint: {config["AzureOpenAI:Endpoint"]}")
print(f"Azure OpenAI API Key: {config["AzureOpenAI:Key"]}")
print(f"Azure OpenAI Model Name: {config["AzureOpenAI:ModelName"]}")
```
